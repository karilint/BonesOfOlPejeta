```python
import pandas as pd
import numpy as np

summary = a2_df_bone_census_ol_pejeta_western.copy()

to_rank = (
    summary
    .drop(index="Column_Total", errors="ignore")
    .drop(columns=["Row_Total"], errors="ignore")
)

ranks_wide = to_rank.rank(axis=0, method='min', ascending=False)

# 1) Reciprocal-rank weights (higher rank -> smaller weight)
recip = (1 / ranks_wide).astype(float).fillna(0)

# 2) Normalize within each year so column weights sum to 1 (makes years comparable)
recip_norm = recip.div(recip.sum(axis=0).replace(0, np.nan), axis=1).fillna(0)

# 3) A cross-year dominance score per species (equal weight to each year)
species_recip_score = recip_norm.mean(axis=1).sort_values(ascending=False)

# Optional: “Borda-like” linear rank weights, also normalised per year
max_rank_per_year = ranks_wide.max(skipna=True)
borda = (max_rank_per_year - ranks_wide + 1)  # top rank gets largest weight
borda_norm = borda.div(borda.sum(axis=0).replace(0, np.nan), axis=1).fillna(0)
species_borda_score = borda_norm.mean(axis=1).sort_values(ascending=False)

borda.head()
```

### What the Borda score represents

The Borda score is a cross-year summary of relative rank dominance. For each year, taxa are ranked by abundance, converted to linear rank weights so that the highest-ranked taxon receives the largest weight, and then normalized within that year. The final Borda score for a taxon is the mean of those normalized yearly weights across all available years.

A **high Borda score** therefore indicates that a taxon was repeatedly among the most abundant components of the death assemblage through time, whereas a **low Borda score** indicates that it was usually rare, inconsistently present, or ranked near the bottom of the assemblage. Because the score is rank-based, it emphasizes relative position within each year rather than absolute specimen counts.

In the western death assemblage, the highest Borda scores belong to **Equus burchellii** (0.1361), **Bovidae (medium)** (0.1000), **Bovidae (large)** (0.0831), **Giraffa camelopardalis** (0.0643), and **Phacochoerus africanus** (0.0615). This indicates that zebra and several medium-to-large herbivore categories dominate the western bone assemblage across years. At the low end, **Hyaenidae**, **Panthera leo**, **Chlorocebus pygerythrus**, **Dendrohyrax arboreus**, and **Alcelaphus buselaphus** have the smallest Borda scores, indicating rare or irregular representation.

As with the eastern bone assemblage, some years contain very few observations, so the Borda score is best interpreted as a summary of repeated rank prominence rather than a precise estimate of abundance through time.

```python
def diversity_metrics(abund: pd.DataFrame) -> pd.DataFrame:
    # proportions per year
    col_sums = abund.sum(axis=0)
    prop = abund.div(col_sums.replace(0, np.nan), axis=1)

    # richness (Hill N0)
    S = (abund > 0).sum(axis=0)

    # Shannon (H'), Hill N1 = exp(H')
    H = -(prop * np.log(prop.replace(0, np.nan))).sum(axis=0).fillna(0)
    N1 = np.exp(H)

    # Simpson D = sum p^2, Hill N2 = 1/D
    D = (prop ** 2).sum(axis=0).fillna(0)
    N2 = (1 / D.replace(0, np.nan)).fillna(0)

    # Evenness: Pielou J' and Hill evenness (N1/S)
    J = (H / np.log(S.replace(0, np.nan))).fillna(0)
    E_hill = (N1 / S.replace(0, np.nan)).fillna(0)

    out = pd.DataFrame({
        "Richness_S": S,
        "Shannon_H'": H,
        "Hill_N1": N1,
        "Simpson_D": D,
        "Hill_N2": N2,
        "Pielou_J'": J,
        "Hill_evenness_N1_over_S": E_hill,
        "Total_Abundance": col_sums
    })
    out.index.name = "Year"
    return out

metrics = diversity_metrics(abund)
metrics.round(3).head(50)
```

### How to read the diversity metrics table

The `metrics` table summarizes several complementary aspects of annual death-assemblage structure. `Richness_S` is the number of taxa recorded in a given year. `Shannon_H'` increases as assemblages become both richer and more even, and `Hill_N1` converts that value into the effective number of common taxa.

`Simpson_D` increases when a small subset of taxa dominates the assemblage, whereas `Hill_N2 = 1/D` expresses the effective number of highly abundant taxa. `Pielou_J'` and `Hill_evenness_N1_over_S` summarize evenness, with higher values indicating more equal representation among the taxa present. `Total_Abundance` records the total number of observations in a year and is especially important here because several bone-sample years are very small.

For the western bone assemblage, richness ranges from **1 taxon in 2005** to **12 taxa in 2015**. Diversity peaks in **2012**, which has the maximum `Shannon_H'` (2.203), `Hill_N1` (9.049), and `Hill_N2` (7.579), indicating the broadest and most even assemblage in the series. In contrast, **2005** consists of a single recorded taxon and therefore has the minimum possible diversity values and the maximum `Simpson_D` (1.000). `Total_Abundance` ranges from **1 in 2005** to **37 in 2015**, so temporal diversity patterns should be interpreted with the underlying sample size in mind.

The evenness metrics again need to be read carefully. `Pielou_J'` is effectively **1.000 in 2008**, indicating a very even distribution among the taxa present that year, whereas `Hill_evenness_N1_over_S` is highest in **2005** only because a single taxon is present. The lowest `Hill_evenness_N1_over_S` occurs in **2007** (0.532), indicating the strongest dominance among the years with more than one recorded taxon.


```python
import matplotlib.pyplot as plt

def whittaker_plot(abund: pd.DataFrame, years=None, log_y=True):
    """
    Draws rank–abundance curves. One curve per year.
    """
    if years is None:
        years = list(abund.columns)

    plt.figure(figsize=(7, 5))
    for yr in years:
        s = abund[yr]
        s = s[s > 0].sort_values(ascending=False)  # positive abundances only
        if s.empty:
            continue
        ranks = np.arange(1, len(s) + 1)
        plt.plot(ranks, s.values, marker="o", label=str(yr))

    plt.xlabel("Rank (most → least abundant)")
    plt.ylabel("Raw count")
    if log_y:
        plt.yscale("log")  # optional log scaling
    plt.title("Rank–Abundance (Whittaker) Plot")
    plt.legend(title="Year")
    plt.tight_layout()
    plt.show()

# Example: plot a few years (or omit 'years=' to plot all)
#whittaker_plot(abund, years=[2005, 2009, 2017, 2023], log_y=True)
#whittaker_plot(prop*100, years=[2009, 2023], log_y=False)
#whittaker_plot(abund, years=[2008, 2019], log_y=True)
#whittaker_plot(abund, years=[2009, 2010, 2019, 2023], log_y=True)
whittaker_plot(abund, years=[2001, 2009, 2010, 2018, 2023], log_y=False)
```

### What the Whittaker plot shows

The Whittaker, or rank-abundance, plot provides a visual summary of dominance structure for selected years using raw counts on a linear y-axis. The length of the curve reflects richness, whereas the steepness of the decline reflects evenness and the absolute separation among the highest-ranked taxa: flatter curves indicate a more even assemblage, and steeper curves indicate stronger dominance by a few taxa. For bone assemblages, these plots should always be read together with sample size, because a very short curve may reflect limited sampling as much as ecological simplicity.

In the western dead assemblage, **2001** should appear relatively short but fairly even, with three leading taxa each contributing **25%** of observations. **2009** and **2010** remain short and moderately steep, with zebra or medium bovids occupying the top rank. **2018** should be one of the longer curves because it contains nine taxa, and dominance is less extreme than in some earlier years. **2023** should reduce to a single-point curve because only **Syncerus caffer** is present, indicating complete dominance in that very small sample rather than a diverse assemblage.
